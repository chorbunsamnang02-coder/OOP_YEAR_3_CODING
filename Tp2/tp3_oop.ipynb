```python
#ex1 
class Greeting:
    def __init__(self , name):
        self.name=name

    def say_hello(self):
        print("hello" , self.name)

input_name=str(input("Please enter your name :"))
greeting = Greeting(input_name)
greeting.say_hello()
```

```python
#ex2
#Write a Python program that reads the width and height of a rectangle and calculates its perimeter and area.
class rectangle:
    def __init__(self, width , height):
        self.wid= width
        self.hei=height
    def perimeter(self):
        return 2*(self.wid + self.hei)
    def area(self):
        return self.hei* self.wid
Height = int(input(" Please enter height :"))
width =int(input("Enter width :"))
Recipe= rectangle(Height , width)
print("Perimeter:", Recipe.perimeter())
print("Area:", Recipe.area())

```

```python
#ex3
#equation calculator
class Equation_Calulator:
    def __init__(self , y , z):
        self.y = y
        self.z=z
    def calculate_x(self):
        x=(self.y*self.z)/(self.y+self.z)
        return x

    def display_result(self):
        if self.y ==0 or self.z ==0 or self.y + self.z==0:
            return "Error: Division by zero"
        else:
            x=self.calculate_x()
            return "The value of x is: ", self.calculate_x()    
y = float(input("Enter the value of y: "))
z = float(input("Enter the value of z: "))
calculator = Equation_Calulator(y, z)
print(calculator.display_result())
```

```python
#exercise 4
class hundredsCounter:
    def __init__(self , number):
        if number<0:
            raise ValueError("error negative number are not allow")
        self.number=number
    def count_hundred(self):
        
            return self.number//100

if __name__=="__main__":
    try:
        userinput=int(input("Please enter non negative number :"))
        counter=hundredsCounter(userinput)
        print(f"The number of hundreds in {userinput} is {counter.count_hundred()}")
    except ValueError as e:
        print(e)
```
